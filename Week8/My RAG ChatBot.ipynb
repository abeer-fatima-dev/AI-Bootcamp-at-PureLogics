```python
%%writefile app.py

import os
import streamlit as st

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

# CONFIGURATION

PDF_PATH = "/content/crime-and-punishment.pdf"

GOOGLE_API_KEY = os.environ.get("GOOGLE_API_KEY")

# PAGE CONFIG
st.set_page_config(
    page_title="Crime and Punishment RAG",
    page_icon="ðŸ“š",
    layout="centered"
)

st.title("Crime and Punishment â€” RAG Assistant")

st.caption(
    "Ask questions about Dostoevsky's Crime and Punishment."
)

# CHECK API KEY
if not GOOGLE_API_KEY:

    st.error(
        "GOOGLE_API_KEY is not configured."
    )

    st.stop()
# CHECK PDF

if not os.path.exists(PDF_PATH):

    st.error(
        f"Book PDF not found at: {PDF_PATH}"
    )

    st.stop()

# LOAD PDF + CREATE VECTOR STORE

@st.cache_resource
def create_vectorstore():

    st.info("Loading Crime and Punishment and creating embeddings...")

    # Load PDF
    loader = PyPDFLoader(PDF_PATH)

    documents = loader.load()

    # Split into chunks
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=100
    )

    chunks = splitter.split_documents(documents)

    # Embedding model
    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )

    # Create Chroma vector store
    vectorstore = Chroma.from_documents(
        documents=chunks,
        embedding=embeddings,
        collection_name="crime_and_punishment"
    )

    return vectorstore


vectorstore = create_vectorstore()

# RETRIEVER
retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={
        "k": 5
    }
)
# GEMINI
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    temperature=0.2,
    google_api_key=GOOGLE_API_KEY
)

# RAG PROMPT
prompt = ChatPromptTemplate.from_template("""

You are a helpful RAG assistant for Fyodor Dostoevsky's
Crime and Punishment.

Answer the user's question using ONLY the retrieved context
from the book.

Rules:

1. Do not invent information.
2. Do not use information that is not supported by the context.
3. If the answer cannot be found in the retrieved context,
   say:

"I could not find enough information in the retrieved text
to answer this question."

4. Give a clear and concise answer.
5. Do not mention that you are an AI.
6. Do not mention the retrieval process unless asked.

Retrieved Context:
------------------
{context}
------------------

Question:
{question}

Answer:
""")

# EXTRACT TEXT FROM GEMINI RESPONSE
def extract_response_text(response):

    # LangChain AIMessage normally stores the answer in .content
    if hasattr(response, "content"):

        content = response.content

    else:

        content = response

    # Normal string response
    if isinstance(content, str):

        return content

    # Structured Gemini response

    if isinstance(content, list):

        text_parts = []

        for item in content:

            if isinstance(item, dict):

                if item.get("type") == "text":

                    text = item.get("text", "")

                    if text:

                        text_parts.append(text)

        if text_parts:

            return "\n".join(text_parts)

    # Fallback

        return str(content)

# RAG FUNCTION
def answer_question(question):

    # 1. Retrieve relevant documents
    docs = retriever.invoke(question)

    # 2. Build context
    context = "\n\n".join(
        doc.page_content
        for doc in docs
    )

    # 3. Create prompt
    messages = prompt.invoke({
        "context": context,
        "question": question
    })

    # 4. Ask Gemini
    response = llm.invoke(messages)

    # 5. Extract clean answer
    answer = extract_response_text(response)

    return answer, docs

# STREAMLIT UI
question = st.text_area(
    "Ask a question",
    placeholder="Example: Who is Sonia?",
    height=100
)

# ASK BUTTON
if st.button("Ask", type="primary"):

    if not question.strip():

        st.warning(
            "Please enter a question first."
        )

    else:

        with st.spinner(
            "Searching Crime and Punishment..."
        ):

            answer, retrieved_docs = answer_question(
                question
            )

        # Display answer
        st.subheader("Answer")

        st.write(answer)

        # Display retrieved sources
        with st.expander(
            "View Retrieved Context"
        ):

            for i, doc in enumerate(
                retrieved_docs,
                start=1
            ):

                st.markdown(
                    f"**Retrieved Passage {i}**"
                )

                st.write(
                    doc.page_content
                )

                # Page metadata
                if "page" in doc.metadata:

                    st.caption(
                        f"PDF Page: {doc.metadata['page'] + 1}"
                    )

                st.divider()
```

```python
import os
import streamlit as st

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

# CONFIGURATION
PDF_PATH = "/content/crime-and-punishment.pdf"

# Make sure this environment variable exists in Colab
GOOGLE_API_KEY = os.environ.get("GOOGLE_API_KEYS")

# PAGE CONFIG
st.set_page_config(
    page_title="Crime and Punishment RAG",
    page_icon="ðŸ“š",
    layout="centered"
)

st.title("Crime and Punishment â€” RAG Assistant")

st.caption(
    "Multi-turn RAG assistant with conversation memory."
)

# CHECK API KEY
if not GOOGLE_API_KEY:
    st.error(
        "GOOGLE_API_KEY is not configured. "
        "Please add your Gemini API key before running the app."
    )
    st.stop()

# STEP 2: INITIALIZE CHAT HISTORY
# This history exists only while the current Streamlit
# session is active. It is NOT stored in a database.

if "messages" not in st.session_state:
    st.session_state.messages = []

# LOAD PDF + CREATE VECTOR STORE
@st.cache_resource
def create_vectorstore():

    loader = PyPDFLoader(PDF_PATH)

    documents = loader.load()

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=100
    )

    chunks = splitter.split_documents(documents)

    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )

    vectorstore = Chroma.from_documents(
        documents=chunks,
        embedding=embeddings,
        collection_name="crime_and_punishment"
    )

    return vectorstore


with st.spinner("Loading Crime and Punishment..."):
    vectorstore = create_vectorstore()

# RETRIEVER
retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 5}
)

# GEMINI
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    temperature=0.2,
    google_api_key=GOOGLE_API_KEY
)

# DISPLAY PREVIOUS CONVERSATION
for message in st.session_state.messages:

    with st.chat_message(message["role"]):
        st.markdown(message["content"])


# RAG + CONVERSATION FUNCTION
def answer_question(question):

    # Retrieve relevant book content
    docs = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content
        for doc in docs
    )

    # Convert previous conversation into text
    conversation_history = ""

    for message in st.session_state.messages:

        role = message["role"].upper()

        conversation_history += (
            f"{role}: {message['content']}\n"
        )

   # Prompt
    prompt = ChatPromptTemplate.from_template(
        """
You are a helpful RAG assistant for
Crime and Punishment by Fyodor Dostoevsky.

Your job is to answer questions using the provided
book context.

IMPORTANT RULES:

1. Use the retrieved book context as the primary source.
2. Use conversation history to understand references
   such as "he", "she", "that character", "what happened next",
   or follow-up questions.
3. Do not invent information.
4. If the answer cannot be supported by the retrieved
   context, say:

"I could not find enough information in the book to answer this question."

5. Keep the answer clear and concise.

CONVERSATION HISTORY:
{history}

RETRIEVED BOOK CONTEXT:
{context}

CURRENT QUESTION:
{question}

ANSWER:
"""
    )

    messages = prompt.invoke({
        "history": conversation_history,
        "context": context,
        "question": question
    })

    response = llm.invoke(messages)

    return response.content

# CHAT INPUT
question = st.chat_input(
    "Ask something about Crime and Punishment..."
)

# PROCESS USER QUESTION
if question:

    # Display user message
    with st.chat_message("user"):
        st.markdown(question)

    # Save user message in session memory
    st.session_state.messages.append({
        "role": "user",
        "content": question
    })

    # Generate answer
    with st.chat_message("assistant"):

        with st.spinner("Searching the book..."):

            answer = answer_question(question)

        st.markdown(answer)

    # Save AI response in session memory
    st.session_state.messages.append({
        "role": "assistant",
        "content": answer
    })

# SIDEBAR
with st.sidebar:

    st.header("Conversation")

    st.write(
        f"Messages in memory: "
        f"{len(st.session_state.messages)}"
    )

    if st.button("Clear Conversation"):

        st.session_state.messages = []

        st.rerun()
```

```python
app_code = r'''
import os
import streamlit as st

from supabase import create_client

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

# CONFIGURATION
PDF_PATH = "/content/crime-and-punishment.pdf"

COLLECTION_NAME = "crime_and_punishment_step3"

CHROMA_PATH = "/content/chroma_db"


# LOAD ENVIRONMENT VARIABLES
GOOGLE_API_KEY = os.environ.get("GOOGLE_API_KEY")
SUPABASE_URL = os.environ.get("SUPABASE_URL")
SUPABASE_KEY = os.environ.get("SUPABASE_KEY")


if not GOOGLE_API_KEY:
    st.error("GOOGLE_API_KEY is missing.")
    st.stop()

if not SUPABASE_URL:
    st.error("SUPABASE_URL is missing.")
    st.stop()

if not SUPABASE_KEY:
    st.error("SUPABASE_KEY is missing.")
    st.stop()

# STREAMLIT PAGE
st.set_page_config(
    page_title="Crime and Punishment RAG",
    page_icon="ðŸ“š",
    layout="centered"
)

st.title("Crime and Punishment RAG Assistant")

st.caption(
    "Persistent multi-turn RAG chatbot using Gemini, "
    "Chroma and Supabase."
)

# CHECK PDF
if not os.path.exists(PDF_PATH):

    st.error(
        f"PDF not found at: {PDF_PATH}"
    )

    st.info(
        "Upload crime-and-punishment.pdf to /content "
        "before starting the Streamlit application."
    )

    st.stop()

# SUPABASE CLIENT
supabase = create_client(
    SUPABASE_URL,
    SUPABASE_KEY
)

# CREATE / LOAD VECTOR STORE
@st.cache_resource
def create_vectorstore():

    print("Loading Crime and Punishment...")

    loader = PyPDFLoader(PDF_PATH)

    documents = loader.load()

    print(f"Loaded {len(documents)} pages.")

    # Chunking
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=100
    )

    chunks = splitter.split_documents(documents)

    print(f"Created {len(chunks)} chunks.")

    # Embeddings
    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )
    # Chroma
    vectorstore = Chroma(
        collection_name=COLLECTION_NAME,
        embedding_function=embeddings,
        persist_directory=CHROMA_PATH
    )

    # Check if collection is empty
    try:
        collection_count = vectorstore._collection.count()
    except Exception:
        collection_count = 0

    # Only add documents if Chroma is empty
    if collection_count == 0:

        print("Chroma collection is empty.")
        print("Adding documents...")

        vectorstore.add_documents(chunks)

        print("Documents added to Chroma.")

    else:

        print(
            f"Existing Chroma collection found "
            f"with {collection_count} documents."
        )

    return vectorstore


with st.spinner(
    "Loading Crime and Punishment vector store..."
):

    vectorstore = create_vectorstore()

# RETRIEVER
retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={
        "k": 5
    }
)

# GEMINI
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    temperature=0.2,
    api_key=GOOGLE_API_KEY
)

# DATABASE FUNCTIONS
def get_messages():

    try:

        result = (
            supabase
            .table("chats")
            .select(
                "id,role,message,created_at"
            )
            .order(
                "created_at",
                desc=False
            )
            .execute()
        )

        return result.data

    except Exception as e:

        st.error(
            f"Could not load chat history: {e}"
        )

        return []


def save_message(role, message):

    try:

        supabase.table("chats").insert({
            "role": role,
            "message": message
        }).execute()

    except Exception as e:

        st.error(
            f"Could not save message: {e}"
        )


def clear_messages():

    try:

        supabase \
            .table("chats") \
            .delete() \
            .gt("id", 0) \
            .execute()

        return True

    except Exception as e:

        st.error(
            f"Could not clear chat history: {e}"
        )

        return False

# LOAD PERSISTENT CHAT HISTORY
chat_history = get_messages()

# SIDEBAR
with st.sidebar:

    st.header("Chat Information")

    st.write(
        f"Messages stored in Supabase: "
        f"{len(chat_history)}"
    )

    st.divider()

    if st.button(
        "Clear Chat History",
        use_container_width=True
    ):

        if clear_messages():

            st.success(
                "Chat history cleared."
            )

            st.rerun()

# DISPLAY OLD CHAT HISTORY
for message in chat_history:

    role = message["role"]

    if role == "ai":
        display_role = "assistant"
    else:
        display_role = "user"

    with st.chat_message(display_role):

        st.markdown(
            message["message"]
        )

# RAG PROMPT
prompt = ChatPromptTemplate.from_template(
    """
You are a RAG assistant for the book
"Crime and Punishment" by Fyodor Dostoevsky.

Your job is to answer the user's question using
the retrieved context from the book and the previous
conversation.

IMPORTANT RULES:

1. Use the retrieved book context as the primary source.

2. Use conversation history to understand follow-up
questions.

3. Do not invent facts.

4. Do not claim information that is not supported
by the retrieved context.

5. If the book context does not contain enough
information to answer the question, say:

"I could not find enough information in the book
to answer this question."

6. Keep the answer clear and reasonably concise.

7. If the user asks a follow-up question such as
"What about her?", use the conversation history
to understand who "her" refers to.

------------------------------------------------------------
CONVERSATION HISTORY
------------------------------------------------------------

{history}

------------------------------------------------------------
RETRIEVED BOOK CONTEXT
------------------------------------------------------------

{context}

------------------------------------------------------------
CURRENT QUESTION
------------------------------------------------------------

{question}

------------------------------------------------------------
ANSWER
------------------------------------------------------------
"""
)

# RAG FUNCTION
def answer_question(question):

    # Retrieve relevant book chunks

    docs = retriever.invoke(question)

    # Build context
    context = "\n\n".join(
        doc.page_content
        for doc in docs
    )

    # Build conversation history
    history_text = ""

    for message in chat_history:

        role = message["role"]

        if role == "user":

            history_text += (
                f"USER: {message['message']}\n"
            )

        else:

            history_text += (
                f"AI: {message['message']}\n"
            )

    # Create prompt
   messages = prompt.invoke({
        "history": history_text,
        "context": context,
        "question": question
    })

    # Ask Gemini
   response = llm.invoke(messages)

    answer = response.content

    # Handle structured Gemini responses
    if isinstance(answer, list):

        text_parts = []

        for item in answer:

            if isinstance(item, dict):

                if item.get("type") == "text":

                    text_parts.append(
                        item.get("text", "")
                    )

            elif isinstance(item, str):

                text_parts.append(item)

        answer = "\n".join(text_parts)

    return str(answer)

# STREAMLIT CHAT INPUT
question = st.chat_input(
    "Ask a question about Crime and Punishment..."
)


# PROCESS NEW QUESTION
if question:

    # Show user message
    with st.chat_message("user"):

        st.markdown(question)

    # Save user message

    save_message(
        "user",
        question
    )

    # Generate AI response
    with st.chat_message("assistant"):

        with st.spinner(
            "Searching the book and generating an answer..."
        ):

            answer = answer_question(question)

        st.markdown(answer)

    # Save AI response

    save_message(
        "ai",
        answer
    )

    # Refresh Streamlit

    st.rerun()

# FOOTER
st.divider()

st.caption(
    "Conversation history is persisted in Supabase. "
    "Book knowledge is retrieved from the Chroma vector store."
)
'''

with open("/content/app.py", "w", encoding="utf-8") as f:
    f.write(app_code)

print("âœ… app.py created successfully!")
```

```python
# Evaluate Crime and Punishment RAG using RAGAS

# 1. INSTALL RAGAS

!pip install -q -U ragas

# 2. IMPORTS
import pandas as pd
from ragas import EvaluationDataset, evaluate
from ragas.llms import LangchainLLMWrapper

from ragas.metrics import (
    LLMContextRecall,
    Faithfulness,
    FactualCorrectness
)

# 3. CREATE TEST QUESTIONS
# IMPORTANT:
# For Context Recall and Factual Correctness, we need
# reference/ground-truth answers.
# These answers should be based on Crime and Punishment.

test_data = [

    {
        "user_input": "Who is Raskolnikov?",
        "reference":
            "Rodion Raskolnikov is the central character of "
            "Crime and Punishment. He is a former student who "
            "commits a murder and struggles with guilt, "
            "morality, and the consequences of his actions."
    },

    {
        "user_input": "Who is Sonia Marmeladova?",
        "reference":
            "Sonia Marmeladova is the daughter of Marmeladov. "
            "She is forced into prostitution to support her family "
            "and becomes an important source of compassion and "
            "moral support for Raskolnikov."
    },

    {
        "user_input": "Why did Raskolnikov kill the old woman?",
        "reference":
            "Raskolnikov kills Alyona Ivanovna because he believes "
            "that extraordinary people may be justified in "
            "transgressing moral laws for a supposedly greater "
            "purpose. His theory is connected to his desire to "
            "test whether he himself is such a person."
    },

    {
        "user_input": "Who is Porfiry Petrovich?",
        "reference":
            "Porfiry Petrovich is the investigating magistrate "
            "who investigates the murder of Alyona Ivanovna. "
            "He becomes increasingly suspicious of Raskolnikov "
            "and uses psychological pressure to expose the truth."
    },

    {
        "user_input": "What happens to Raskolnikov at the end of the novel?",
        "reference":
            "Raskolnikov confesses to his crime and is sent to "
            "Siberia for punishment. Sonia follows him and remains "
            "supportive. His imprisonment begins a process of "
            "moral and spiritual transformation."
    },

    {
        "user_input": "What is the relationship between Sonia and Raskolnikov?",
        "reference":
            "Sonia becomes closely connected to Raskolnikov and "
            "provides him with compassion, emotional support, and "
            "moral guidance. Their relationship becomes important "
            "to Raskolnikov's spiritual transformation."
    },

    {
        "user_input": "Who is Svidrigailov?",
        "reference":
            "Arkady Svidrigailov is a wealthy and morally ambiguous "
            "character connected to Dunya and Raskolnikov. His "
            "actions and character provide a contrast to Raskolnikov "
            "and contribute to the novel's exploration of morality."
    },

    {
        "user_input": "What is Raskolnikov's theory about extraordinary people?",
        "reference":
            "Raskolnikov's theory suggests that extraordinary people "
            "may have the right to violate ordinary moral laws if "
            "their actions are directed toward a greater purpose."
    },

    {
        "user_input": "Who was Alyona Ivanovna?",
        "reference":
            "Alyona Ivanovna was an elderly pawnbroker whom "
            "Raskolnikov murders at the beginning of the novel."
    },

    {
        "user_input": "How does Sonia influence Raskolnikov?",
        "reference":
            "Sonia influences Raskolnikov through her compassion, "
            "faith, loyalty, and willingness to remain with him "
            "despite his crime. Her support contributes to his "
            "eventual moral and spiritual transformation."
    }
]


print("Number of test questions:", len(test_data))


# 4. RUN YOUR RAG SYSTEM ON EACH QUESTION
evaluation_rows = []


for i, item in enumerate(test_data):

    question = item["user_input"]
    reference = item["reference"]

    print("=" * 70)
    print(f"Question {i + 1}/{len(test_data)}")
    print(question)

    try:

        # Retrieve documents
        docs = retriever.invoke(question)

        retrieved_contexts = [
            doc.page_content
            for doc in docs
        ]
     # Generate answer using YOUR RAG application
        answer = answer_question(question)
    # Save evaluation sample
        evaluation_rows.append({

            "user_input": question,

            "retrieved_contexts":
                retrieved_contexts,

            "response":
                str(answer),

            "reference":
                reference
        })

        print("\nGenerated answer:")
        print(answer)

    except Exception as e:

        print("\nERROR:")
        print(e)

# 5. CHECK GENERATED DATASET
print("\n")
print("=" * 70)
print("RAG EVALUATION DATASET")
print("=" * 70)

print(
    "Successful samples:",
    len(evaluation_rows)
)

# 6. CONVERT TO RAGAS EvaluationDataset
evaluation_dataset = EvaluationDataset.from_list(
    evaluation_rows
)

print("\nRagas dataset created successfully.")

print(
    "Features:",
    evaluation_dataset.features()
)

# 7. CREATE EVALUATOR LLM
# We use Gemini to judge our RAG application's answers.
evaluator_llm = LangchainLLMWrapper(llm)

# 8. SELECT RAGAS METRICS
metrics = [

    # Checks whether retrieved context supports
    # the information required by the reference answer.
    LLMContextRecall(
        llm=evaluator_llm
    ),

    # Checks whether the generated answer is
    # supported by the retrieved context.
    Faithfulness(
        llm=evaluator_llm
    ),

    # Checks factual correctness against
    # the reference answer.
    FactualCorrectness(
        llm=evaluator_llm
    )
]

# 9. RUN EVALUATION
print("\n")
print("=" * 70)
print("STARTING RAGAS EVALUATION")
print("=" * 70)

results = evaluate(

    dataset=evaluation_dataset,

    metrics=metrics,

    llm=evaluator_llm
)

# 10. DISPLAY RESULTS
print("\n")
print("=" * 70)
print("RAGAS RESULTS")
print("=" * 70)

print(results)


# 11. CONVERT RESULTS TO DATAFRAME
results_df = results.to_pandas()

print("\n")
print("=" * 70)
print("PER-QUESTION RESULTS")
print("=" * 70)

display(results_df)

# 12. CALCULATE AVERAGE SCORES
score_columns = [
    column
    for column in [
        "context_recall",
        "faithfulness",
        "factual_correctness"
    ]
    if column in results_df.columns
]


print("\n")
print("=" * 70)
print("AVERAGE SCORES")
print("=" * 70)

average_scores = results_df[score_columns].mean()

display(
    average_scores.to_frame(
        name="Average Score"
    )
)

# 13. INTERPRET THE RESULTS
print("\n")
print("=" * 70)
print("INTERPRETATION")
print("=" * 70)

for metric_name, score in average_scores.items():

    print(
        f"{metric_name}: {score:.3f}"
    )

    if score >= 0.90:

        print(
            "  Excellent performance."
        )

    elif score >= 0.75:

        print(
            "  Good performance, but there is room for improvement."
        )

    elif score >= 0.60:

        print(
            "  Moderate performance. Retrieval or generation "
            "should be improved."
        )

    else:

        print(
            "  Poor performance. Investigate retrieval, "
            "chunking, prompts, or the LLM."
        )

# 14. SAVE RESULTS
results_df.to_csv(
    "/content/ragas_results.csv",
    index=False
)

print(
    "\nResults saved to:"
    "/content/ragas_results.csv"
)

# 15. SAVE TEST DATASET
pd.DataFrame(
    test_data
).to_json(
    "/content/ragas_test_dataset.json",
    orient="records",
    indent=2
)

print(
    "Test dataset saved to:"
    "/content/ragas_test_dataset.json"
)


# 16. FINAL SUMMARY
print("\n")
print("=" * 70)
print("STEP 4 COMPLETED")
print("=" * 70)

print("""
Your Crime and Punishment RAG application has now been
evaluated using Ragas.

Metrics evaluated:

1. Context Recall
   -> Did retrieval find the information needed?

2. Faithfulness
   -> Is the generated answer supported by the retrieved
      book context?

3. Factual Correctness
   -> Is the generated answer factually consistent with
      the expected/reference answer?

Use the scores to decide what to improve next:
- chunk size
- chunk overlap
- number of retrieved documents (k)
- embedding model
- prompt
- Gemini temperature
- retrieval strategy
""")
```
