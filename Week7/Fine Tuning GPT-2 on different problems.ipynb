## 2. Quick-start with Pipelines

The `pipeline` API wraps tokenization, model loading, and inference in one object.

### 2.1 Sentiment Analysis

```python
from transformers import pipeline

# 1. Load a sentiment-analysis pipeline (defaults to a small model)
sentiment = pipeline("sentiment-analysis")

# 2. Run inference
examples = [
    "Hugging Face makes NLP super accessible!",
    "I dislike bugs in my code..."
]
results = sentiment(examples)
for text, res in zip(examples, results):
    print(f"{text!r:50} → label={res['label']}, score={res['score']:.3f}")
```

### 2.2 Text Generation

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
prompt = "Once upon a time"
out = generator(prompt, max_length=30, num_return_sequences=1)
print(out[0]["generated_text"])
```

## Technical Approaches to Fine-Tuning

### Hierarchical Domain Adaptation

When dealing with related but distinct domains, hierarchical fine-tuning approaches offer advantages over single-domain methods.

Research shows that representing domains hierarchically allows models to capture both domain-specific and general-domain information[^3]. By using adapter layers to represent nodes in a hierarchical structure, models can selectively share representations between related domains while avoiding negative transfer from unrelated domains[^3].

This approach is particularly valuable when fine-tuning for multiple related domains, such as hotel reviews and restaurant reviews, which share some linguistic features but retain domain-specific characteristics[^3]. The hierarchical structure allows upper nodes to be updated more frequently, encoding more general knowledge, while leaf nodes capture domain-specific information[^3].

### Implementation Considerations

Implementing GPT-2 fine-tuning for custom applications involves several critical steps that ensure successful model development.

The process typically begins with installing dependencies and importing necessary libraries, including PyTorch and Hugging Face Transformers[^2]. Data processing involves splitting datasets into training and validation sets (typically 90% for training, 10% for validation)[^2]. Tokenization using the GPT-2 tokenizer prepares input and target sequences with appropriate padding and truncation options[^2].

Training configurations must specify parameters like output directory, number of training epochs, batch size, evaluation strategy, and logging frequency[^2]. After fine-tuning, the model can be saved and deployed for inference, with consideration given to computational resources like GPU availability for optimal performance[^2].

# **Problem 1: Sentiment Analysis on Movie Reviews**

````python
# ==========================================================
# Sentiment Analysis on Movie Reviews — Fine-tune GPT-2
# Dataset: IMDb (Hugging Face Datasets)
#   !pip install -U -q datasets huggingface_hub fsspec transformers accelerate evaluate


import numpy as np
import torch

# Force the datasets library to skip its torchvision integration.
# torchvision isn't used anywhere in this text-classification script, but
# datasets auto-detects it and tries to import VideoReader from it, which
# breaks on newer torchvision versions where VideoReader was removed.
# Setting this flag before importing datasets avoids that broken code path.
import datasets.config
datasets.config.TORCHVISION_AVAILABLE = False

from datasets import load_dataset
from transformers import (
    GPT2Tokenizer,
    GPT2ForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding,
)
import evaluate

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# ==========================================================
# 1. Load the IMDb dataset
# ==========================================================
raw_datasets = load_dataset("stanfordnlp/imdb")
print(raw_datasets)


train_dataset = raw_datasets["train"].shuffle(seed=42).select(range(5000))
test_dataset = raw_datasets["test"].shuffle(seed=42).select(range(2000))

# ==========================================================
# 2. Tokenizer — GPT-2 has no pad token by default, so we add one
# ==========================================================
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token  # GPT-2 fix: reuse eos as pad

def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=256,  # movie reviews can be long; 256 keeps training fast
    )

tokenized_train = train_dataset.map(tokenize_function, batched=True)
tokenized_test = test_dataset.map(tokenize_function, batched=True)

# Remove the raw text column, rename label -> labels (Trainer expects "labels")
tokenized_train = tokenized_train.remove_columns(["text"])
tokenized_test = tokenized_test.remove_columns(["text"])
tokenized_train = tokenized_train.rename_column("label", "labels")
tokenized_test = tokenized_test.rename_column("label", "labels")
tokenized_train.set_format("torch")
tokenized_test.set_format("torch")

data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

# ==========================================================
# 3. Load GPT-2 for sequence classification (binary: pos/neg)
# ==========================================================
model = GPT2ForSequenceClassification.from_pretrained(
    "gpt2",
    num_labels=2,
    id2label={0: "NEGATIVE", 1: "POSITIVE"},
    label2id={"NEGATIVE": 0, "POSITIVE": 1},
)
# GPT-2's classification head needs to know the pad token id
model.config.pad_token_id = tokenizer.pad_token_id
model.to(device)

# ==========================================================
# 4. Metrics
# ==========================================================
accuracy_metric = evaluate.load("accuracy")
f1_metric = evaluate.load("f1")

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    acc = accuracy_metric.compute(predictions=predictions, references=labels)
    f1 = f1_metric.compute(predictions=predictions, references=labels)
    return {"accuracy": acc["accuracy"], "f1": f1["f1"]}

# ==========================================================
# 5. Training arguments and Trainer
# ==========================================================
training_args = TrainingArguments(
    output_dir="./gpt2-imdb-sentiment",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_steps=50,
    load_best_model_at_end=True,
    metric_for_best_model="accuracy",
    fp16=torch.cuda.is_available(),  # mixed precision if on GPU
    report_to="none",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_test,
    processing_class=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
)

# ==========================================================
# 6. Fine-tune
# ==========================================================
trainer.train()

# ==========================================================
# 7. Evaluate on the test set
# ==========================================================
eval_results = trainer.evaluate()
print("Final evaluation results:", eval_results)

# ==========================================================
# 8. Save the fine-tuned model
# ==========================================================
trainer.save_model("./gpt2-imdb-sentiment/final")
tokenizer.save_pretrained("./gpt2-imdb-sentiment/final")

# ==========================================================
# 9. Run inference on new, unseen reviews
# ==========================================================
def predict_sentiment(text, model, tokenizer, device):
    model.eval()
    inputs = tokenizer(
        text, return_tensors="pt", truncation=True, max_length=256
    ).to(device)
    with torch.no_grad():
        logits = model(**inputs).logits
    probs = torch.softmax(logits, dim=-1).squeeze()
    pred_id = int(torch.argmax(probs))
    label = model.config.id2label[pred_id]
    confidence = float(probs[pred_id])
    return label, confidence

sample_reviews = [
    "This movie was an absolute masterpiece, the acting and cinematography blew me away.",
    "Terrible plot, wooden acting, and a complete waste of two hours.",
    "It was okay, not great but not bad either, some parts dragged on.",
]

print("\n--- Sample predictions ---")
for review in sample_reviews:
    label, conf = predict_sentiment(review, model, tokenizer, device)
    print(f"Review: {review[:60]}...")
    print(f"  -> Predicted: {label} (confidence: {conf:.3f})\n")
````

# Problem 2:Domain-Specific Text Generation (e.g., Harry Potter Fan Fiction)

````python
import urllib.request
url = "https://www.gutenberg.org/cache/epub/55/pg55.txt"
raw = urllib.request.urlopen(url).read().decode("utf-8")
start = raw.find("*** START OF")
start = raw.find("\n", start) + 1
end = raw.find("*** END OF")
story_text = raw[start:end].strip()
with open("my_corpus.txt", "w", encoding="utf-8") as f:
    f.write(story_text)
print(f"Saved {len(story_text)} characters to my_corpus.txt")
````

````python
# ==========================================================
# Domain-Specific Text Generation — Fine-tune GPT-2 for language modeling
#
# This script fine-tunes GPT-2 on a plain-text corpus you provide, so it
# generates new text in that corpus's style (fan fiction, a specific
# author's voice, a genre, etc.).

#   !pip install -U -q datasets huggingface_hub fsspec transformers accelerate evaluate
import math
import torch
import datasets.config
datasets.config.TORCHVISION_AVAILABLE = False

from datasets import load_dataset
from transformers import (
    GPT2Tokenizer,
    GPT2LMHeadModel,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling,
)

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# 0. Point this at your own corpus file
CORPUS_PATH = "my_corpus.txt"  # <-- upload your .txt file in Colab's file browser and update this path

# 1. Load the text file as a Hugging Face dataset
raw_datasets = load_dataset("text", data_files={"train": CORPUS_PATH})

# Split off a small validation set so we can track overfitting
split_datasets = raw_datasets["train"].train_test_split(test_size=0.05, seed=42)
print(split_datasets)

# 2. Tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token

BLOCK_SIZE = 256  # sequence length per training example

def tokenize_function(examples):
    return tokenizer(examples["text"])

tokenized_datasets = split_datasets.map(
    tokenize_function, batched=True, remove_columns=["text"]
)

# Concatenate all tokenized text, then chop into fixed-length blocks.
# This is the standard approach for language-model fine-tuning: rather than
# padding every short line, we pack tokens end-to-end so training is efficient.
def group_texts(examples):
    concatenated = sum(examples["input_ids"], [])
    total_length = (len(concatenated) // BLOCK_SIZE) * BLOCK_SIZE
    result_ids = [
        concatenated[i : i + BLOCK_SIZE] for i in range(0, total_length, BLOCK_SIZE)
    ]
    return {"input_ids": result_ids, "labels": [ids.copy() for ids in result_ids]}

lm_datasets = tokenized_datasets.map(
    group_texts,
    batched=True,
    remove_columns=tokenized_datasets["train"].column_names,
)
lm_datasets.set_format("torch")

# For causal LM, the collator just needs to know we're NOT doing masked LM
data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

# 3. Load GPT-2 for language modeling
model = GPT2LMHeadModel.from_pretrained("gpt2")
model.to(device)

# 4. Training arguments and Trainer
training_args = TrainingArguments(
    output_dir="./gpt2-domain-generator",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=5e-5,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    num_train_epochs=5,
    weight_decay=0.01,
    logging_steps=50,
    load_best_model_at_end=True,
    fp16=torch.cuda.is_available(),
    report_to="none",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=lm_datasets["train"],
    eval_dataset=lm_datasets["test"],
    processing_class=tokenizer,
    data_collator=data_collator,
)

# 5. Fine-tune
trainer.train()

# 6. Evaluate — report perplexity, the standard LM quality metric
eval_results = trainer.evaluate()
perplexity = math.exp(eval_results["eval_loss"])
print(f"Final eval loss: {eval_results['eval_loss']:.4f}")
print(f"Perplexity: {perplexity:.2f}  (lower is better)")

# 7. Save the fine-tuned model
trainer.save_model("./gpt2-domain-generator/final")
tokenizer.save_pretrained("./gpt2-domain-generator/final")

# 8. Generate new text in the fine-tuned style
def generate_text(
    prompt,
    model,
    tokenizer,
    device,
    max_new_tokens=150,
    temperature=0.9,
    top_p=0.92,
    top_k=50,
    num_return_sequences=1,
):
    model.eval()
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            num_return_sequences=num_return_sequences,
            pad_token_id=tokenizer.eos_token_id,
        )
    return [tokenizer.decode(o, skip_special_tokens=True) for o in outputs]

prompt = "The castle grounds were quiet that morning, until"
generated = generate_text(prompt, model, tokenizer, device, num_return_sequences=2)

print("\n--- Generated samples ---")
for i, text in enumerate(generated, 1):
    print(f"\nSample {i}:\n{text}")
````

# Problem 3: FAQ Answer Generation for a Public Dataset

````python
# ==========================================================
# FAQ / Q&A Generation — Fine-tune GPT-2 on question-answer pairs
#
# Fine-tunes GPT-2 so that given "Question: <text>\nAnswer:" it generates
# a plausible answer, learned from a dataset of real Q&A pairs.
#
# This script uses the COVID-19 QA dataset (deepset/covid_qa_deepset) from
# the Hugging Face Hub as the public data source — real questions and
# answers extracted from official COVID-19 research papers. Swap the
# dataset name in step 1 to use a different domain (e.g. a Stack Overflow
# Q&A dump) if you'd prefer.
#   !pip install -U -q datasets huggingface_hub fsspec transformers accelerate evaluate
# ==========================================================

import math
import torch

# Skip datasets' broken torchvision integration (unrelated to this task,
# but avoids a common ImportError in Colab environments).
import datasets.config
datasets.config.TORCHVISION_AVAILABLE = False

from datasets import load_dataset
from transformers import (
    GPT2Tokenizer,
    GPT2LMHeadModel,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling,
)

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# 1. Load a question-answer dataset
raw_datasets = load_dataset("deepset/covid_qa_deepset")
print(raw_datasets)
print(raw_datasets["train"][0])  # inspect the fields: question, answers, context, ...

# The 'answers' field is a dict like {"text": [...], "answer_start": [...]}.
# We pull out the first answer text for each example.
def extract_answer(example):
    example["answer_text"] = example["answers"]["text"][0]
    return example

raw_datasets = raw_datasets.map(extract_answer)

# Keep only the columns we need, split into train/validation
raw_datasets = raw_datasets["train"].select_columns(["question", "answer_text"])
split_datasets = raw_datasets.train_test_split(test_size=0.05, seed=42)
print(split_datasets)

# 2. Format each example as a single prompt-response training string
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token

def format_qa(example):
    text = f"Question: {example['question']}\nAnswer: {example['answer_text']}{tokenizer.eos_token}"
    return {"text": text}

formatted_datasets = split_datasets.map(format_qa, remove_columns=["question", "answer_text"])

# 3. Tokenize
MAX_LENGTH = 256

def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=MAX_LENGTH,
        padding="max_length",
    )

tokenized_datasets = formatted_datasets.map(
    tokenize_function, batched=True, remove_columns=["text"]
)
tokenized_datasets.set_format("torch")

# Causal LM collator — labels are the input_ids themselves, shifted
# internally by the model during loss computation
data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

# ==========================================================
# 4. Load GPT-2 for language modeling
# ==========================================================
model = GPT2LMHeadModel.from_pretrained("gpt2")
model.to(device)

# ==========================================================
# 5. Training arguments and Trainer
# ==========================================================
training_args = TrainingArguments(
    output_dir="./gpt2-faq-generator",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=5e-5,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_steps=50,
    load_best_model_at_end=True,
    fp16=torch.cuda.is_available(),
    report_to="none",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["test"],
    processing_class=tokenizer,
    data_collator=data_collator,
)

# ==========================================================
# 6. Fine-tune
# ==========================================================
trainer.train()

# ==========================================================
# 7. Evaluate — perplexity
# ==========================================================
eval_results = trainer.evaluate()
perplexity = math.exp(eval_results["eval_loss"])
print(f"Final eval loss: {eval_results['eval_loss']:.4f}")
print(f"Perplexity: {perplexity:.2f}  (lower is better)")

# ==========================================================
# 8. Save the fine-tuned model
# ==========================================================
trainer.save_model("./gpt2-faq-generator/final")
tokenizer.save_pretrained("./gpt2-faq-generator/final")

# ==========================================================
# 9. Ask new questions
# ==========================================================
def answer_question(
    question,
    model,
    tokenizer,
    device,
    max_new_tokens=80,
    temperature=0.7,
    top_p=0.9,
):
    model.eval()
    prompt = f"Question: {question}\nAnswer:"
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=temperature,
            top_p=top_p,
            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )
    full_text = tokenizer.decode(output[0], skip_special_tokens=True)
    # Return just the generated answer portion
    return full_text.split("Answer:", 1)[-1].strip()

sample_questions = [
    "What are the common symptoms of COVID-19?",
    "How does COVID-19 spread between people?",
    "What is the incubation period for COVID-19?",
]

print("\n--- Sample generated answers ---")
for q in sample_questions:
    answer = answer_question(q, model, tokenizer, device)
    print(f"\nQ: {q}\nA: {answer}")
````

# Problem 4:  Headline Generation for News Articles

````python
# ==========================================================
# News Headline Generation — Fine-tune GPT-2 on article-headline pairs
#
# Fine-tunes GPT-2 so that given "Article: <snippet>\nHeadline:" it
# generates a concise headline, learned from the CNN/DailyMail dataset.
#
# NOTE on the data: CNN/DailyMail's "highlights" field is actually a short
# multi-sentence bullet-point summary, not a single-line headline in the
# strict sense. It's the standard public dataset used for this kind of
# task and works well for "concise headline/summary generation" - just be
# aware the outputs will look more like 1-3 sentence summaries than
# newspaper-style headlines. If you want stricter single-line headlines,
# you can truncate 'highlights' to its first sentence (shown below).

!pip install -U -q datasets huggingface_hub fsspec transformers accelerate evaluate rouge_score
# ==========================================================

import math
import torch
import datasets.config
datasets.config.TORCHVISION_AVAILABLE = False

from datasets import load_dataset
from transformers import (
    GPT2Tokenizer,
    GPT2LMHeadModel,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling,
)
import evaluate

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# ==========================================================
# 1. Load the CNN/DailyMail dataset
# ==========================================================
raw_datasets = load_dataset("abisee/cnn_dailymail", "3.0.0")
print(raw_datasets)
print(raw_datasets["train"][0]["article"][:300])
print("---")
print(raw_datasets["train"][0]["highlights"])

train_dataset = raw_datasets["train"].shuffle(seed=42).select(range(8000))
val_dataset = raw_datasets["validation"].shuffle(seed=42).select(range(1000))

# ==========================================================
# 2. Format each example as a single prompt-response training string
# ==========================================================
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token

ARTICLE_SNIPPET_CHARS = 800  # how much of the article to feed as context

def format_example(example):
    snippet = example["article"][:ARTICLE_SNIPPET_CHARS]
    headline = example["highlights"].strip()
    text = f"Article: {snippet}\nHeadline: {headline}{tokenizer.eos_token}"
    return {"text": text, "headline_ref": headline, "article_snippet": snippet}

train_formatted = train_dataset.map(format_example, remove_columns=raw_datasets["train"].column_names)
val_formatted = val_dataset.map(format_example, remove_columns=raw_datasets["train"].column_names)

val_references = val_formatted["headline_ref"]
val_article_snippets = val_formatted["article_snippet"]

# ==========================================================
# 3. Tokenize
# ==========================================================
MAX_LENGTH = 384

def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=MAX_LENGTH,
        padding="max_length",
    )

train_tokenized = train_formatted.map(
    tokenize_function, batched=True, remove_columns=["text", "headline_ref", "article_snippet"]
)
val_tokenized = val_formatted.map(
    tokenize_function, batched=True, remove_columns=["text", "headline_ref", "article_snippet"]
)
train_tokenized.set_format("torch")
val_tokenized.set_format("torch")

data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

# ==========================================================
# 4. Load GPT-2 for language modeling
# ==========================================================
model = GPT2LMHeadModel.from_pretrained("gpt2")
model.to(device)

# ==========================================================
# 5. Training arguments and Trainer
# ==========================================================
training_args = TrainingArguments(
    output_dir="./gpt2-headline-generator",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=5e-5,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_steps=50,
    load_best_model_at_end=True,
    fp16=torch.cuda.is_available(),
    report_to="none",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokenized,
    eval_dataset=val_tokenized,
    processing_class=tokenizer,
    data_collator=data_collator,
)

# ==========================================================
# 6. Fine-tune
# ==========================================================
trainer.train()

# ==========================================================
# 7. Evaluate — perplexity (LM quality)
# ==========================================================
eval_results = trainer.evaluate()
perplexity = math.exp(eval_results["eval_loss"])
print(f"Final eval loss: {eval_results['eval_loss']:.4f}")
print(f"Perplexity: {perplexity:.2f}  (lower is better)")

# ==========================================================
# 8. Save the fine-tuned model
# ==========================================================
trainer.save_model("./gpt2-headline-generator/final")
tokenizer.save_pretrained("./gpt2-headline-generator/final")

# ==========================================================
# 9. Generate headlines and score them with ROUGE
# ==========================================================
def generate_headline(
    article_snippet,
    model,
    tokenizer,
    device,
    max_new_tokens=40,
    temperature=0.7,
    top_p=0.9,
):
    model.eval()
    prompt = f"Article: {article_snippet}\nHeadline:"
    inputs = tokenizer(
        prompt, return_tensors="pt", truncation=True, max_length=MAX_LENGTH
    ).to(device)
    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=temperature,
            top_p=top_p,
            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )
    full_text = tokenizer.decode(output[0], skip_special_tokens=True)
    return full_text.split("Headline:", 1)[-1].strip()

# Generate headlines for a handful of validation articles and compute ROUGE
rouge_metric = evaluate.load("rouge")

NUM_EVAL_SAMPLES = 50
predictions = []
references = []

for i in range(min(NUM_EVAL_SAMPLES, len(val_article_snippets))):
    pred = generate_headline(val_article_snippets[i], model, tokenizer, device)
    predictions.append(pred)
    references.append(val_references[i])

rouge_results = rouge_metric.compute(predictions=predictions, references=references)
print("\n--- ROUGE scores on validation sample ---")
for key, value in rouge_results.items():
    print(f"{key}: {value:.4f}")

print("\n--- Example generated headlines vs. reference ---")
for i in range(5):
    print(f"\nGenerated: {predictions[i]}")
    print(f"Reference: {references[i]}")
````

````python
# ==========================================================
# Poetry Generation — Fine-tune GPT-2
# ==========================================================

!pip install -U -q datasets huggingface_hub fsspec transformers accelerate

import os
import math
import torch

from datasets import load_dataset
from transformers import (
    GPT2Tokenizer,
    GPT2LMHeadModel,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling,
)

# ==========================================================
# 1. Check GPU
# ==========================================================

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# ==========================================================
# 2. Set CORPUS PATH
# ==========================================================

CORPUS_PATH = "/content/poetry_corpus.txt"

# Check whether file exists
if not os.path.exists(CORPUS_PATH):
    raise FileNotFoundError(
        f"Poetry file not found at: {CORPUS_PATH}"
    )

print("Corpus found:", CORPUS_PATH)

# ==========================================================
# 3. Load poetry dataset
# ==========================================================

raw_dataset = load_dataset(
    "text",
    data_files={"train": CORPUS_PATH}
)

print(raw_dataset)

# Use a train/test split
split_datasets = raw_dataset["train"].train_test_split(
    test_size=0.2,
    seed=42
)

print(split_datasets)

# ==========================================================
# 4. Load GPT-2 tokenizer
# ==========================================================

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

# GPT-2 does not have a default padding token
# So we use EOS token as padding token
tokenizer.pad_token = tokenizer.eos_token

# Smaller block size because our dataset is small
BLOCK_SIZE = 32

# ==========================================================
# 5. Tokenization
# ==========================================================

def tokenize_function(examples):

    texts = [
        text + tokenizer.eos_token
        for text in examples["text"]
    ]

    return tokenizer(
        texts,
        truncation=False
    )


tokenized_datasets = split_datasets.map(
    tokenize_function,
    batched=True,
    remove_columns=["text"]
)

# ==========================================================
# 6. Group tokens into blocks
# ==========================================================

def group_texts(examples):

    # Combine all tokenized examples
    concatenated = sum(examples["input_ids"], [])

    # Keep only complete blocks
    total_length = (
        len(concatenated) // BLOCK_SIZE
    ) * BLOCK_SIZE

    input_ids = [
        concatenated[i:i + BLOCK_SIZE]
        for i in range(
            0,
            total_length,
            BLOCK_SIZE
        )
    ]

    return {
        "input_ids": input_ids,
        "labels": [ids.copy() for ids in input_ids]
    }


lm_datasets = tokenized_datasets.map(
    group_texts,
    batched=True,
    remove_columns=tokenized_datasets["train"].column_names,
)

lm_datasets.set_format("torch")

print("Training examples:", len(lm_datasets["train"]))
print("Testing examples:", len(lm_datasets["test"]))

# ==========================================================
# 7. Data Collator
# ==========================================================

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

# ==========================================================
# 8. Load GPT-2
# ==========================================================

model = GPT2LMHeadModel.from_pretrained("gpt2")

model.to(device)

# ==========================================================
# 9. Training arguments
# ==========================================================

training_args = TrainingArguments(
    output_dir="/content/gpt2-poetry-generator",

    # Number of epochs
    num_train_epochs=3,

    # Small dataset → small batch size
    per_device_train_batch_size=2,
    per_device_eval_batch_size=2,

    learning_rate=5e-5,

    weight_decay=0.01,

    logging_steps=1,

    eval_strategy="epoch",
    save_strategy="epoch",

    fp16=torch.cuda.is_available(),

    report_to="none"
)

# ==========================================================
# 10. Create Trainer
# ==========================================================

trainer = Trainer(
    model=model,
    args=training_args,

    train_dataset=lm_datasets["train"],
    eval_dataset=lm_datasets["test"],

    processing_class=tokenizer,

    data_collator=data_collator,
)

# ==========================================================
# 11. Fine-tune GPT-2
# ==========================================================

print("\nStarting GPT-2 fine-tuning...\n")

trainer.train()

print("\nTraining completed!")


# ==========================================================
# Evaluation
# ==========================================================

eval_results = trainer.evaluate()

print("\nEvaluation results:")
print(eval_results)

if "eval_loss" in eval_results:
    eval_loss = eval_results["eval_loss"]

    try:
        perplexity = math.exp(eval_loss)
    except OverflowError:
        perplexity = float("inf")

    print("Eval Loss:", eval_loss)
    print("Perplexity:", perplexity)
else:
    print("Evaluation loss is not available.")
    print("This can happen because the current poetry dataset is very small.")

# ==========================================================
# 13. Save model
# ==========================================================

FINAL_PATH = "/content/gpt2-poetry-generator/final"

trainer.save_model(FINAL_PATH)
tokenizer.save_pretrained(FINAL_PATH)

print("\nModel saved at:")
print(FINAL_PATH)

# ==========================================================
# 14. Generate poetry
# ==========================================================

def generate_poem(
    prompt,
    max_new_tokens=50,
    temperature=0.8,
    top_p=0.92,
    top_k=50
):

    model.eval()

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(device)

    with torch.no_grad():

        outputs = model.generate(
            **inputs,

            max_new_tokens=max_new_tokens,

            do_sample=True,

            temperature=temperature,

            top_p=top_p,

            top_k=top_k,

            pad_token_id=tokenizer.eos_token_id,

            eos_token_id=tokenizer.eos_token_id
        )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )


# ==========================================================
# 15. Test prompts
# ==========================================================

prompts = [
    "The moon",
    "The morning sun",
    "The river"
]

print("\n==============================")
print("GENERATED POETRY")
print("==============================")

for prompt in prompts:

    result = generate_poem(
        prompt,
        max_new_tokens=50,
        temperature=0.8,
        top_p=0.92,
        top_k=50
    )

    print("\nPrompt:", prompt)
    print(result)
    print("-" * 50)
````
